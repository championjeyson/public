# Extract polygon pixel spectra for radiometry evaluation

This notebook extracts **all valid 11-band pixel spectra** inside the spectrally
constant-feature polygons at **Westham** and **Brunswick** for the May 4 and
May 18, 2026 flights.

It compares:

- DLS and NoDLS imagery;
- May 4 and May 18 imagery;
- within-polygon radiometric variability;
- polygon-mean spectra;
- DLS-to-NoDLS ratios and differences;
- agreement between paired products using bias, MAE, RMSE, and $R^2$.

## Important assumptions

1. The files named `10Band` contain **11 source bands**.
2. Source band 5 is PAN, following the convention used in the existing repository notebooks:
   `B1, B2, B3, B4, PAN, B5, B6, B7, B8, B9, B10`.
3. PAN is retained in the extracted tables, but excluded from wavelength-based spectral
   plots by default because the existing project wavelength list contains the 10
   narrow spectral bands only.
4. Raster values are treated as reflectance-like values. Dataset scale/offset metadata
   are applied only when they are non-default.
5. The extraction CRS defaults to `EPSG:6339`, matching the workaround in the existing
   imagery-comparison notebook. Set `ASSUMED_RASTER_CRS = None` to use each raster's
   reported CRS directly.

> No SAS URL or token is stored in this notebook.

## 3. Analysis parameters and band definitions

````python
ALL_TOUCHED = False

# The existing comparison notebook uses EPSG:6339 because the rasters report a
# compound horizontal + vertical CRS. Set this to None if the reported raster CRS
# should be used without substitution.
ASSUMED_RASTER_CRS = None #"EPSG:3157"

# Require every band in a pixel spectrum to be valid.
REQUIRE_ALL_BANDS_VALID = True

# Apply raster scale/offset metadata only if it differs from scale=1, offset=0.
APPLY_NONDEFAULT_SCALE_OFFSET = True

SAVE_OUTPUTS = True
OVERWRITE_OUTPUTS = True

OUTPUT_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "imagery"
    / "radiometry_eval"
    / "polygon_pixel_spectra"
)

# Eleven source bands: the fifth source band is PAN.
SOURCE_BAND_NUMBERS = list(range(1, 12))
SOURCE_BAND_NAMES = [
    "B1", "B2", "B3", "B4", "PAN",
    "B5", "B6", "B7", "B8", "B9", "B10",
]

# The 10 narrow-band center wavelengths used in the existing repository notebook.
SPECTRAL_WAVELENGTHS_NM = {
    "B1": 444,
    "B2": 475,
    "B3": 531,
    "B4": 560,
    "B5": 650,
    "B6": 668,
    "B7": 705,
    "B8": 717,
    "B9": 740,
    "B10": 842,
}

BAND_METADATA = pd.DataFrame({
    "source_band": SOURCE_BAND_NUMBERS,
    "band_name": SOURCE_BAND_NAMES,
})
BAND_METADATA["wavelength_nm"] = BAND_METADATA["band_name"].map(
    SPECTRAL_WAVELENGTHS_NM
)
BAND_METADATA["is_pan"] = BAND_METADATA["band_name"].eq("PAN")

display(BAND_METADATA)
````

## 5. Inspect rasters and polygon files

````python
def inspect_raster(path, test_window_size=16):
    with rasterio.open(path) as src:
        width = min(test_window_size, src.width)
        height = min(test_window_size, src.height)
        test = src.read(
            1,
            window=rasterio.windows.Window(0, 0, width, height),
            masked=True,
        )

        descriptions = list(src.descriptions)
        scales = list(src.scales)
        offsets = list(src.offsets)

        return {
            "path": path,
            "driver": src.driver,
            "width": src.width,
            "height": src.height,
            "count": src.count,
            "dtypes": src.dtypes,
            "nodata": src.nodata,
            "reported_crs": str(src.crs),
            "assumed_extraction_crs": ASSUMED_RASTER_CRS,
            "transform": src.transform,
            "bounds": tuple(src.bounds),
            "descriptions": descriptions,
            "scales": scales,
            "offsets": offsets,
            "test_read_shape": test.shape,
            "test_read_successful": True,
        }


raster_info_records = []
for (flight_date, site, dls_mode), path in RASTER_PATHS.items():
    record = inspect_raster(path)
    record.update({
        "flight_date": flight_date,
        "site": site,
        "dls_mode": dls_mode,
    })
    raster_info_records.append(record)

raster_info = (
    pd.DataFrame(raster_info_records)
    .set_index(["flight_date", "site", "dls_mode"])
    .sort_index()
)

display(raster_info.T)

if not raster_info["count"].eq(11).all():
    raise ValueError(
        "Every configured radiometry raster was expected to contain 11 bands. "
        f"Observed counts: {sorted(raster_info['count'].unique())}"
    )
````

````python
def choose_feature_id_field(gdf):
    candidates = [
        "feature_id", "id", "ID", "name", "Name",
        "label", "Label", "description", "Description",
        "Descrip", "class", "Class",
    ]
    for candidate in candidates:
        if candidate in gdf.columns:
            return candidate
    return None


polygon_layers = {}

for site, path in POLYGON_PATHS.items():
    gdf = gpd.read_file(path)

    if gdf.crs is None:
        raise ValueError(f"{site} polygons have no CRS: {path}")
    if gdf.geometry.isna().any():
        raise ValueError(f"{site} polygons contain missing geometries.")
    if (~gdf.geometry.is_valid).any():
        warnings.warn(f"{site} contains invalid geometries; applying buffer(0).")
        gdf["geometry"] = gdf.geometry.buffer(0)

    id_field = choose_feature_id_field(gdf)
    gdf = gdf.reset_index(drop=False).rename(columns={"index": "source_feature_index"})

    if id_field is None:
        gdf["feature_id"] = [
            f"{site}_{i:03d}" for i in range(1, len(gdf) + 1)
        ]
    else:
        gdf["feature_id"] = (
            gdf[id_field].astype("string").fillna("").str.strip()
        )
        missing_id = gdf["feature_id"].eq("")
        gdf.loc[missing_id, "feature_id"] = [
            f"{site}_{i:03d}"
            for i in gdf.loc[missing_id, "source_feature_index"]
        ]

    gdf["site"] = site
    polygon_layers[site] = gdf

    print(
        f"{site}: {len(gdf)} polygons; CRS={gdf.crs}; "
        f"feature ID source={id_field or 'generated'}"
    )
    display(gdf.drop(columns="geometry").head())
````

## 6. Extraction functions

The pixel table is written in **wide format**:

- one row per valid raster pixel inside a polygon;
- one column per source band;
- source row/column and pixel-center coordinates;
- date, site, DLS mode, and polygon ID.

The polygon summary is written in **long format** with one row per
image × polygon × band.

````python
def extraction_crs(src):
    if ASSUMED_RASTER_CRS is not None:
        return ASSUMED_RASTER_CRS
    if src.crs is None:
        raise ValueError(
            "Raster has no CRS and ASSUMED_RASTER_CRS is None."
        )
    return src.crs


def nondefault_scale_offset(src):
    scales = np.asarray(src.scales, dtype=float)
    offsets = np.asarray(src.offsets, dtype=float)
    nondefault = (
        not np.allclose(scales, 1.0)
        or not np.allclose(offsets, 0.0)
    )
    return scales, offsets, nondefault


def extract_polygon_pixels(
    raster_path,
    polygon_gdf,
    *,
    flight_date,
    site,
    dls_mode,
    all_touched=ALL_TOUCHED,
    require_all_bands_valid=REQUIRE_ALL_BANDS_VALID,
):
    pixel_frames = []
    feature_records = []

    with rasterio.open(raster_path) as src:
        if src.count != 11:
            raise ValueError(
                f"Expected 11 bands but found {src.count}: {raster_path}"
            )

        target_crs = extraction_crs(src)
        polygons = polygon_gdf.to_crs(target_crs).copy()

        scales, offsets, has_nondefault_scale_offset = nondefault_scale_offset(src)
        apply_scale_offset = (
            APPLY_NONDEFAULT_SCALE_OFFSET and has_nondefault_scale_offset
        )

        flight_date_day = int(flight_date.split('-')[-1])
        for _, feature in polygons.iterrows():
            feature_id = str(feature["feature_id"])
            feature_day = int(feature["day"]) if feature["day"] is not None and not np.isnan(feature["day"]) else None
            # print(feature_id, feature_day)
            if feature_day is not None and feature_day != flight_date_day:
                # print(f"Skipping feature {feature_id} due to day mismatch.")
                continue
            geom = feature.geometry

            try:
                window = geometry_window(
                    src,
                    [mapping(geom)],
                    pad_x=0,
                    pad_y=0,
                    boundless=False,
                )
            except rasterio.errors.WindowError:
                warnings.warn(
                    f"No raster overlap: {flight_date}, {site}, "
                    f"{dls_mode}, {feature_id}"
                )
                feature_records.append({
                    "flight_date": flight_date,
                    "site": site,
                    "dls_mode": dls_mode,
                    "feature_id": feature_id,
                    "pixel_count": 0,
                    "status": "no_overlap",
                })
                continue

            data = src.read(
                indexes=SOURCE_BAND_NUMBERS,
                window=window,
                masked=True,
            )
            transform = src.window_transform(window)

            inside = geometry_mask(
                [mapping(geom)],
                out_shape=(int(window.height), int(window.width)),
                transform=transform,
                invert=True,
                all_touched=all_touched,
            )

            values = np.ma.getdata(data).astype("float64", copy=False)
            band_mask = np.ma.getmaskarray(data)

            if apply_scale_offset:
                values = (
                    values * scales[:, None, None]
                    + offsets[:, None, None]
                )

            finite = np.isfinite(values)

            if require_all_bands_valid:
                valid = (
                    inside
                    & (~band_mask).all(axis=0)
                    & finite.all(axis=0)
                )
            else:
                valid = inside & ((~band_mask) & finite).any(axis=0)

            local_rows, local_cols = np.where(valid)
            pixel_count = len(local_rows)

            if pixel_count == 0:
                warnings.warn(
                    f"No valid pixels: {flight_date}, {site}, "
                    f"{dls_mode}, {feature_id}"
                )
                feature_records.append({
                    "flight_date": flight_date,
                    "site": site,
                    "dls_mode": dls_mode,
                    "feature_id": feature_id,
                    "pixel_count": 0,
                    "status": "no_valid_pixels",
                })
                continue

            global_rows = local_rows + int(window.row_off)
            global_cols = local_cols + int(window.col_off)

            xs, ys = xy(
                src.transform,
                global_rows,
                global_cols,
                offset="center",
            )

            pixel_values = values[:, local_rows, local_cols].T

            pixel_df = pd.DataFrame(
                pixel_values,
                columns=SOURCE_BAND_NAMES,
            )
            pixel_df.insert(0, "y", np.asarray(ys))
            pixel_df.insert(0, "x", np.asarray(xs))
            pixel_df.insert(0, "raster_col", global_cols)
            pixel_df.insert(0, "raster_row", global_rows)
            pixel_df.insert(0, "pixel_id", np.arange(pixel_count))
            pixel_df.insert(0, "feature_id", feature_id)
            pixel_df.insert(0, "dls_mode", dls_mode)
            pixel_df.insert(0, "site", site)
            pixel_df.insert(0, "flight_date", flight_date)
            pixel_df["raster_path"] = raster_path
            pixel_df["scale_offset_applied"] = apply_scale_offset

            pixel_frames.append(pixel_df)

            feature_records.append({
                "flight_date": flight_date,
                "site": site,
                "dls_mode": dls_mode,
                "feature_id": feature_id,
                "pixel_count": pixel_count,
                "status": "ok",
                "scale_offset_applied": apply_scale_offset,
            })

    pixels = (
        pd.concat(pixel_frames, ignore_index=True)
        if pixel_frames
        else pd.DataFrame()
    )
    extraction_log = pd.DataFrame(feature_records)
    return pixels, extraction_log


def summarize_pixel_spectra(pixel_df):
    if pixel_df.empty:
        return pd.DataFrame()

    id_columns = [
        "flight_date", "site", "dls_mode", "feature_id",
    ]

    long_pixels = pixel_df.melt(
        id_vars=id_columns + [
            "pixel_id", "raster_row", "raster_col", "x", "y",
        ],
        value_vars=SOURCE_BAND_NAMES,
        var_name="band_name",
        value_name="value",
    )

    summary = (
        long_pixels
        .groupby(id_columns + ["band_name"], observed=True)["value"]
        .agg(
            pixel_count="count",
            mean="mean",
            median="median",
            std="std",
            minimum="min",
            maximum="max",
            q05=lambda x: x.quantile(0.05),
            q25=lambda x: x.quantile(0.25),
            q75=lambda x: x.quantile(0.75),
            q95=lambda x: x.quantile(0.95),
        )
        .reset_index()
    )

    summary["cv_percent"] = np.where(
        summary["mean"].abs() > 0,
        100 * summary["std"] / summary["mean"].abs(),
        np.nan,
    )

    summary = summary.merge(
        BAND_METADATA,
        on="band_name",
        how="left",
        validate="many_to_one",
    )

    return summary
````

## 7. Run extraction and save per-image outputs

````python
if SAVE_OUTPUTS:
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

all_pixel_tables = []
all_summaries = []
all_logs = []

for (flight_date, site, dls_mode), raster_path in sorted(RASTER_PATHS.items()):
    print(f"Extracting: {flight_date} | {site} | {dls_mode}")
    print("Raster:", raster_path)

    pixels, log = extract_polygon_pixels(
        raster_path,
        polygon_layers[site],
        flight_date=flight_date,
        site=site,
        dls_mode=dls_mode,
    )

    # # Compute NDVI 740 (using 668 and 740 nm bands)
    # if not pixels.empty:
    #     pixels["NDVI_740"] = (pixels["B9"] - pixels["B6"]) / (pixels["B9"] + pixels["B6"])

    summary = summarize_pixel_spectra(pixels)

    all_pixel_tables.append(pixels)
    all_summaries.append(summary)
    all_logs.append(log)

    print(
        f"  extracted pixels={len(pixels):,}; "
        f"features with data={(log['status'] == 'ok').sum() if not log.empty else 0}"
    )

    if SAVE_OUTPUTS:
        stem = f"{flight_date}_{site}_{dls_mode}".replace("-", "")
        pixel_path = OUTPUT_DIR / f"{stem}_pixel_spectra.parquet"
        summary_path = OUTPUT_DIR / f"{stem}_polygon_band_summary.parquet"
        log_path = OUTPUT_DIR / f"{stem}_extraction_log.csv"

        if OVERWRITE_OUTPUTS or not pixel_path.exists():
            pixels.to_parquet(pixel_path, index=False)
            summary.to_parquet(summary_path, index=False)
            log.to_csv(log_path, index=False)

pixel_spectra = pd.concat(all_pixel_tables, ignore_index=True)
polygon_band_summary = pd.concat(all_summaries, ignore_index=True)
extraction_log = pd.concat(all_logs, ignore_index=True)

display(extraction_log)
display(polygon_band_summary.head())

if SAVE_OUTPUTS:
    polygon_band_summary.to_parquet(
        OUTPUT_DIR / "all_polygon_band_summaries.parquet",
        index=False,
    )
    extraction_log.to_csv(
        OUTPUT_DIR / "all_extraction_logs.csv",
        index=False,
    )

print("Total extracted pixel spectra:", f"{len(pixel_spectra):,}")
print("Output directory:", OUTPUT_DIR)
````

## 9. Plot polygon-mean spectra: DLS versus NoDLS

````python
N_COLUMNS = 3
FEATURES_PER_FIGURE = 3*N_COLUMNS

DLS_ORDER = ["NoDLS", "DLS"]
DLS_COLORS = {
    "NoDLS": "#0072B2",
    "DLS": "#D55E00",
}


def natural_sort_key(value):
    """Sort IDs such as feature2 before feature10."""
    return [
        int(part) if part.isdigit() else part.lower()
        for part in re.split(r"(\d+)", str(value))
    ]


spectral_summary = polygon_band_summary.loc[
    ~polygon_band_summary["is_pan"] & polygon_band_summary["wavelength_nm"].notna()
].copy()

required_columns = {
    "flight_date",
    "site",
    "dls_mode",
    "feature_id",
    "wavelength_nm",
    "mean",
    "std",
}
missing_columns = required_columns.difference(spectral_summary.columns)
if missing_columns:
    raise KeyError(
        f"polygon_band_summary is missing required columns: {sorted(missing_columns)}"
    )


for flight_date in sorted(spectral_summary["flight_date"].unique()):
    for site in sorted(spectral_summary["site"].unique()):
        image_data = spectral_summary.loc[
            spectral_summary["flight_date"].eq(flight_date)
            & spectral_summary["site"].eq(site)
        ].sort_values(["feature_id", "dls_mode", "wavelength_nm"])

        if image_data.empty:
            continue

        feature_ids = sorted(
            image_data["feature_id"].dropna().astype(str).unique(),
            key=natural_sort_key,
        )

        # Paginate features so a site with many polygons does not create one
        # unmanageably large figure.
        for page_start in range(0, len(feature_ids), FEATURES_PER_FIGURE):
            page_features = feature_ids[page_start : page_start + FEATURES_PER_FIGURE]

            n_features = len(page_features)
            ncols = min(N_COLUMNS, n_features)
            nrows = int(np.ceil(n_features / ncols))

            fig, axes = plt.subplots(
                nrows=nrows,
                ncols=ncols,
                figsize=(5.2 * ncols, 3.8 * nrows),
                sharex=True,
                sharey=False,
                squeeze=False,
            )
            axes = axes.ravel()

            for ax, feature_id in zip(axes, page_features):
                feature_data = image_data.loc[
                    image_data["feature_id"].astype(str).eq(feature_id)
                ]

                for dls_mode in DLS_ORDER:
                    mode_data = feature_data.loc[
                        feature_data["dls_mode"].eq(dls_mode)
                    ].sort_values("wavelength_nm")

                    if mode_data.empty:
                        continue

                    wavelength = mode_data["wavelength_nm"].to_numpy(dtype=float)
                    mean = mode_data["mean"].to_numpy(dtype=float)
                    std = mode_data["std"].fillna(0).to_numpy(dtype=float)

                    color = DLS_COLORS[dls_mode]

                    ax.plot(
                        wavelength,
                        mean,
                        color=color,
                        marker="o",
                        markersize=4,
                        linewidth=2,
                    )
                    ax.fill_between(
                        wavelength,
                        mean - std,
                        mean + std,
                        color=color,
                        alpha=0.18,
                        linewidth=0,
                    )

                # Get description of feature from polygon_layers[site]
                feature_description = polygon_layers[site].loc[
                    polygon_layers[site]["feature_id"].astype(str).eq(feature_id)
                ].iloc[0]["description"]
                ax.set_title(f"{str(feature_id)}: {feature_description.capitalize()}")
                ax.set_xlabel("Center wavelength (nm)")
                ax.set_ylabel("Raster value")
                ax.grid(True, alpha=0.25)

            for ax in axes[n_features:]:
                ax.remove()

            legend_handles = [
                Line2D(
                    [0],
                    [0],
                    color=DLS_COLORS[dls_mode],
                    marker="o",
                    linewidth=2,
                    label=f"{dls_mode}: mean ± pixel SD",
                )
                for dls_mode in DLS_ORDER
            ]

            page_number = page_start // FEATURES_PER_FIGURE + 1
            page_count = int(np.ceil(len(feature_ids) / FEATURES_PER_FIGURE))

            fig.legend(
                handles=legend_handles,
                loc="upper center",
                bbox_to_anchor=(0.5, 0.995),
                ncol=2,
                frameon=True,
            )
            fig.suptitle(
                "Feature spectra: mean ± within-feature pixel SD\n"
                f"{site.title()} | {flight_date} | "
                f"page {page_number}/{page_count}",
                y=1.025,
                fontsize=15,
            )
            fig.tight_layout(rect=(0, 0, 1, 0.95))
            plt.show()
````

## 12. DLS versus NoDLS agreement metrics

````python
def agreement_metrics(group):
    valid = group[["NoDLS", "DLS"]].dropna()

    if valid.empty:
        return pd.Series({
            "n": 0,
            "bias": np.nan,
            "mae": np.nan,
            "rmse": np.nan,
            "r2": np.nan,
            "mean_ratio": np.nan,
        })

    reference = valid["NoDLS"].to_numpy()
    estimate = valid["DLS"].to_numpy()
    residual = estimate - reference

    return pd.Series({
        "n": len(valid),
        "bias": residual.mean(),
        "mae": np.abs(residual).mean(),
        "rmse": np.sqrt(np.mean(residual ** 2)),
        "r2": r2_score(reference, estimate) if len(valid) >= 2 else np.nan,
        "mean_ratio": np.nanmean(
            np.divide(
                estimate,
                reference,
                out=np.full_like(estimate, np.nan, dtype=float),
                where=np.abs(reference) > 0,
            )
        ),
    })


dls_agreement = (
    paired_dls
    .groupby(
        ["flight_date", "site", "band_name", "source_band", "is_pan"],
        observed=True,
    )
    .apply(agreement_metrics)
    .reset_index()
)

display(dls_agreement)

if SAVE_OUTPUTS:
    dls_agreement.to_csv(
        OUTPUT_DIR / "DLS_NoDLS_agreement_metrics.csv",
        index=False,
    )
````

## 16. Fit temporal radiometric correction factors

This section fits linear transformations between the two flight dates using
paired polygon-band means for the same site, DLS mode, feature, and band.

The fitted correction is:

\[
y_{\mathrm{target}} =
\mathrm{gain}\,x_{\mathrm{source}} + \mathrm{offset}
\]

It calculates:

1. **Eleven separate per-band fits**, including PAN.
2. **One combined fit using all 10 narrow spectral bands**, excluding PAN.
3. Both directions:
   - May 4 → May 18
   - May 18 → May 4
4. Separate fits for each site and DLS mode by default.

The reverse direction is fitted independently rather than calculated as the
mathematical inverse of the forward fit. Ordinary least-squares regression
is asymmetric, so independently fitted forward and reverse coefficients will
generally not be exact inverses.

````python
from sklearn.linear_model import LinearRegression

DATE_MAY04 = "2026-05-04"
DATE_MAY18 = "2026-05-18"

TEMPORAL_FIT_DIRECTIONS = [
    (DATE_MAY04, DATE_MAY18),
    (DATE_MAY18, DATE_MAY04),
]

# True: separate correction factors for Westham and Brunswick.
# False: pool both sites into one fit for each DLS mode.
FIT_TEMPORAL_CORRECTIONS_BY_SITE = True

MIN_TEMPORAL_FIT_PAIRS = 3
````

### 16.1 Pair the same feature and band between dates

````python
temporal_pair_index = [
    "site",
    "dls_mode",
    "feature_id",
    "band_name",
    "source_band",
    "is_pan",
]

# wavelength_nm is deliberately not included in the pivot index because
# PAN has wavelength_nm = NaN and would otherwise be dropped.
temporal_paired = (
    polygon_band_summary
    .pivot_table(
        index=temporal_pair_index,
        columns="flight_date",
        values="mean",
        aggfunc="first",
    )
    .reset_index()
)
temporal_paired.columns.name = None

missing_dates = {
    DATE_MAY04,
    DATE_MAY18,
}.difference(temporal_paired.columns)

if missing_dates:
    raise KeyError(
        "The temporal pairing table is missing date columns: "
        f"{sorted(missing_dates)}"
    )

# Add wavelength metadata after pivoting so PAN remains in the table.
temporal_paired = temporal_paired.merge(
    BAND_METADATA[
        [
            "band_name",
            "source_band",
            "wavelength_nm",
        ]
    ],
    on=["band_name", "source_band"],
    how="left",
    validate="many_to_one",
)

temporal_paired["has_both_dates"] = (
    temporal_paired[DATE_MAY04].notna()
    & temporal_paired[DATE_MAY18].notna()
)

temporal_pair_qc = (
    temporal_paired
    .groupby(
        ["site", "dls_mode", "band_name"],
        observed=True,
        dropna=False,
    )
    .agg(
        available_may04=(DATE_MAY04, "count"),
        available_may18=(DATE_MAY18, "count"),
        paired_features=("has_both_dates", "sum"),
    )
    .reset_index()
)

display(temporal_pair_qc)
display(temporal_paired.head())

if SAVE_OUTPUTS:
    temporal_paired.to_parquet(
        OUTPUT_DIR / "temporal_paired_polygon_band_means.parquet",
        index=False,
    )
    temporal_pair_qc.to_csv(
        OUTPUT_DIR / "temporal_pairing_qc.csv",
        index=False,
    )
````

### 16.2 Linear-fit function

````python
def fit_temporal_linear_correction(
    data,
    *,
    source_date,
    target_date,
    minimum_pairs=MIN_TEMPORAL_FIT_PAIRS,
):
    valid = (
        data[[source_date, target_date]]
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
    )

    n = len(valid)
    source_mean = (
        valid[source_date].mean()
        if n else np.nan
    )
    target_mean = (
        valid[target_date].mean()
        if n else np.nan
    )

    if (
        n < minimum_pairs
        or valid[source_date].nunique() < 2
    ):
        return {
            "n_pairs": n,
            "gain": np.nan,
            "offset": np.nan,
            "r2": np.nan,
            "rmse": np.nan,
            "mae": np.nan,
            "bias_after_correction": np.nan,
            "source_mean": source_mean,
            "target_mean": target_mean,
            "fit_status": (
                "insufficient_pairs_or_source_variation"
            ),
        }

    x = valid[source_date].to_numpy(dtype=float)
    y = valid[target_date].to_numpy(dtype=float)

    model = LinearRegression()
    model.fit(x.reshape(-1, 1), y)
    predicted = model.predict(x.reshape(-1, 1))
    residual = predicted - y

    return {
        "n_pairs": n,
        "gain": float(model.coef_[0]),
        "offset": float(model.intercept_),
        "r2": float(r2_score(y, predicted)),
        "rmse": float(
            np.sqrt(np.mean((predicted - y) ** 2))
        ),
        "mae": float(np.mean(np.abs(predicted - y))),
        "bias_after_correction": float(residual.mean()),
        "source_mean": float(source_mean),
        "target_mean": float(target_mean),
        "fit_status": "ok",
    }
````

### 16.3 Fit every band separately, including PAN

````python
temporal_fit_group_columns = ["dls_mode"]

if FIT_TEMPORAL_CORRECTIONS_BY_SITE:
    temporal_fit_group_columns.insert(0, "site")

per_band_group_columns = temporal_fit_group_columns + [
    "band_name",
    "source_band",
    "is_pan",
    "wavelength_nm",
]

per_band_fit_records = []

for group_values, group in temporal_paired.groupby(
    per_band_group_columns,
    observed=True,
    dropna=False,
):
    if not isinstance(group_values, tuple):
        group_values = (group_values,)

    group_metadata = dict(
        zip(per_band_group_columns, group_values)
    )

    for source_date, target_date in TEMPORAL_FIT_DIRECTIONS:
        fit = fit_temporal_linear_correction(
            group,
            source_date=source_date,
            target_date=target_date,
        )

        per_band_fit_records.append({
            **group_metadata,
            "fit_scope": "individual_band",
            "source_date": source_date,
            "target_date": target_date,
            "direction": (
                f"{source_date}_to_{target_date}"
            ),
            **fit,
        })

temporal_per_band_factors = (
    pd.DataFrame(per_band_fit_records)
    .sort_values(
        temporal_fit_group_columns
        + ["source_date", "source_band"]
    )
    .reset_index(drop=True)
)

display(temporal_per_band_factors)
````

### 16.4 Fit all 10 narrow spectral bands together, excluding PAN

````python
ten_band_data = temporal_paired.loc[
    ~temporal_paired["is_pan"]
    & temporal_paired["band_name"].ne("PAN")
].copy()

all_ten_band_fit_records = []

for group_values, group in ten_band_data.groupby(
    temporal_fit_group_columns,
    observed=True,
    dropna=False,
):
    if not isinstance(group_values, tuple):
        group_values = (group_values,)

    group_metadata = dict(
        zip(temporal_fit_group_columns, group_values)
    )

    for source_date, target_date in TEMPORAL_FIT_DIRECTIONS:
        fit = fit_temporal_linear_correction(
            group,
            source_date=source_date,
            target_date=target_date,
        )

        all_ten_band_fit_records.append({
            **group_metadata,
            "band_name": "ALL_10_SPECTRAL_BANDS",
            "source_band": pd.NA,
            "is_pan": False,
            "wavelength_nm": np.nan,
            "fit_scope": "all_10_spectral_bands_pooled",
            "source_date": source_date,
            "target_date": target_date,
            "direction": (
                f"{source_date}_to_{target_date}"
            ),
            **fit,
        })

temporal_all_ten_band_factors = (
    pd.DataFrame(all_ten_band_fit_records)
    .sort_values(
        temporal_fit_group_columns
        + ["source_date"]
    )
    .reset_index(drop=True)
)

display(temporal_all_ten_band_factors)
````

### 16.5 Combined correction-factor table

### 16.7 Observed versus fitted temporal values

````python
def add_temporal_predictions(
        paired_data,
        factor_table,
        *,
        source_date,
        target_date,
        fit_scope,
    ):
        factors = factor_table.loc[
            factor_table["source_date"].eq(source_date)
            & factor_table["target_date"].eq(target_date)
            & factor_table["fit_scope"].eq(fit_scope)
            & factor_table["fit_status"].eq("ok")
        ].copy()

        if fit_scope == "individual_band":
            merge_columns = temporal_fit_group_columns + [
                "band_name",
                "source_band",
                "is_pan",
            ]
            data = paired_data.copy()
        else:
            merge_columns = temporal_fit_group_columns
            data = paired_data.loc[
                ~paired_data["is_pan"]
                & paired_data["band_name"].ne("PAN")
            ].copy()

        predicted = data.merge(
            factors[
                merge_columns + ["gain", "offset"]
            ],
            on=merge_columns,
            how="inner",
            validate="many_to_one",
        )

        predicted["source_value"] = predicted[source_date]
        predicted["target_value"] = predicted[target_date]
        predicted["predicted_target"] = (
            predicted["gain"] * predicted["source_value"]
            + predicted["offset"]
        )
        predicted["source_date"] = source_date
        predicted["target_date"] = target_date
        predicted["direction"] = (
            f"{source_date}_to_{target_date}"
        )
        predicted["fit_scope"] = fit_scope

        return predicted


temporal_predictions = []

for source_date, target_date in TEMPORAL_FIT_DIRECTIONS:
    temporal_predictions.append(
        add_temporal_predictions(
            temporal_paired,
            temporal_correction_factors,
            source_date=source_date,
            target_date=target_date,
            fit_scope="individual_band",
        )
    )
    temporal_predictions.append(
        add_temporal_predictions(
            temporal_paired,
            temporal_correction_factors,
            source_date=source_date,
            target_date=target_date,
            fit_scope="all_10_spectral_bands_pooled",
        )
    )

temporal_predictions = pd.concat(
    temporal_predictions,
    ignore_index=True,
    sort=False,
)

if SAVE_OUTPUTS:
    temporal_predictions.to_parquet(
        OUTPUT_DIR / "temporal_correction_predictions.parquet",
        index=False,
    )

for fit_scope in [
    "individual_band",
    "all_10_spectral_bands_pooled",
]:
    scope_data = temporal_predictions.loc[
        temporal_predictions["fit_scope"].eq(fit_scope)
    ]

    if scope_data.empty:
        continue

    if fit_scope == "individual_band":
        g = sns.relplot(
            data=scope_data,
            x="target_value",
            y="predicted_target",
            hue="dls_mode",
            style="site" if FIT_TEMPORAL_CORRECTIONS_BY_SITE else None,
            col="band_name",
            row="direction",
            col_order=SOURCE_BAND_NAMES,
            kind="scatter",
            facet_kws={"sharex": False, "sharey": False},
            height=3.1,
            aspect=1,
            alpha=0.7,
        )
    else:
        g = sns.relplot(
            data=scope_data,
            x="target_value",
            y="predicted_target",
            hue="dls_mode",
            style="site" if FIT_TEMPORAL_CORRECTIONS_BY_SITE else None,
            col="direction",
            kind="scatter",
            facet_kws={"sharex": False, "sharey": False},
            height=4.5,
            aspect=1,
            alpha=0.7,
        )

    for ax in g.axes.flat:
        xlim = ax.get_xlim()
        ylim = ax.get_ylim()
        low = min(xlim[0], ylim[0])
        high = max(xlim[1], ylim[1])
        ax.plot(
            [low, high],
            [low, high],
            "--",
            color="black",
            linewidth=1,
        )
        ax.set_xlim(low, high)
        ax.set_ylim(low, high)

    g.set_axis_labels(
        "Observed target-date value",
        "Corrected source-date value",
    )
    g.fig.suptitle(
        f"Temporal correction fit: {fit_scope}",
        y=1.02,
        fontsize=14,
    )
    plt.show()
````

### Applying a coefficient

Each output row defines:

```python
corrected_value = gain * source_date_value + offset
```

For example, a row whose direction is `2026-05-04_to_2026-05-18`
converts a May 4 value into the fitted May 18 reference.

The `all_10_spectral_bands_pooled` fit uses every matched feature-band
observation from the ten narrow bands as a separate regression point.
PAN is excluded only from this pooled fit; it remains in the
`individual_band` results.

## Interpretation checklist

For each polygon and band, review:

1. **Pixel count** — unexpectedly low counts can indicate CRS, coverage, nodata,
   or polygon-size issues.
2. **Within-polygon CV** — spectrally constant targets should generally have low
   spatial variability.
3. **DLS/NoDLS ratio** — a consistent offset or gain by band suggests a systematic
   radiometric difference.
4. **DLS versus NoDLS scatter** — departures from the 1:1 line identify bias and
   nonlinearity.
5. **May 4 versus May 18 scatter** — stable features should remain near the 1:1 line
   after accounting for acquisition and processing differences.
6. **PAN behavior** — PAN is retained for QC but should not be interpreted as one of
   the 10 narrow spectral bands.
7. **Polygon geometry** — inspect any feature with zero pixels or an unusual pixel
   count before interpreting its spectra.
